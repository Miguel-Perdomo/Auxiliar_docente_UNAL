Este notebook muestra cómo importar tu script global de estilos `styles.py` desde las subcarpetas del curso.

```
import sys
import os

# Añadir la ruta de la carpeta 'utils' al path de Python (dos niveles arriba)
sys.path.append(os.path.abspath('../../utils'))

import styles
import matplotlib.pyplot as plt
import numpy as np

# Aplicar el estilo para gráficas científicas
# NOTA: Requiere tener TeX Live / MiKTeX instalado en tu sistema local para que text.usetex = True funcione.
styles.paper_style()

# Crear gráfica de prueba
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label='Sen(x)')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud (m)')
plt.title('Ejemplo de Gráfica con LaTeX')
plt.legend()
plt.show()

# Para guardar:
# plt.savefig('../imagenes/grafica1.pdf', bbox_inches='tight')
```
